```python
def rand_mask(bs, sl):
    b = torch.zeros(bs, 1, sl)
    for i in range(bs):
        pad_start = random.randint(0,sl)
        b[i,0,0:pad_start] = 1
    return b
```

```python
def causal_mask(bs,sl):
    b = torch.ones(bs, sl, sl)
    for i in range(bs):
        b[i] = torch.tril(torch.ones((sl, sl)))
    print(b.shape)
    return b
```

```python
#mask=rand_mask(bs, sl2).type(torch.bool).unsqueeze(1)
mask=causal_mask(bs,sl2).type(torch.bool).unsqueeze(1)
printt(mask)

print(mask[0,0,1,4])
```

```python
#print(score)
print(score.masked_fill(~mask, float("-inf")))
```
