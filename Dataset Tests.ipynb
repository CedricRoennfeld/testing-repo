```python
import re

def string_length_and_character_filter(input_string):
    # Remove allowed punctuation and spaces from the input string
    cleaned_input = re.sub(r"[ \!\',;:¡¿\-\.?]", '', input_string)
    
    # Check if the remaining characters are all alphabetic and if the original string has a valid length
    if 8 <= len(input_string) <= 256 and cleaned_input.isalpha():
        return True
    else:
        return False

# Examples
print(string_length_and_character_filter("Hello, world!"))     # Output: True
print(string_length_and_character_filter("What's up!"))        # Output: True
print(string_length_and_character_filter("Hello_world"))       # Output: False
print(string_length_and_character_filter("¡Hola, mundo!"))     # Output: True
print(string_length_and_character_filter("Héllö, Wörld!"))     # Output: True
print(string_length_and_character_filter("Hello@world"))       # Output: False
print(string_length_and_character_filter("Well, why not?"))    # Output: True
print(string_length_and_character_filter("Look: a test."))     # Output: True
print(string_length_and_character_filter("Is this correct?"))  # Output: True
```

```python
import re

def tokenize(seq):
    # Define patterns
    stc = r"[!,;:¡¿.?]"  # Special token characters
    word = r"\b\w+(?:[-']\w+)*\b"  # Match words with internal connectors
    connector = r"[-']"  # Standalone connector (if not part of a word)
    
    # Use regex to capture all valid tokens
    pattern = f"({stc})|({word})|({connector})"
    tokens = [match.group() for match in re.finditer(pattern, seq)]
    
    return tokens

# Example usage:
seq = "Hello, world! I'm Cedric-how are you?"
tokens = tokenize(seq)
print(tokens)
```
